## 문제 2 · 날개가 길면 몸무게도 무거울까? — 관계를 숫자로

`난이도 하` · `예상 20분`

**📖 상황**

- 문제 1에서 "숫자 전에 그림"을 배웠습니다. 이제 관계를 **숫자 하나**로 요약해 봅니다.

- 이번에는 **날개 길이(`flipper_length_mm`)와 몸무게(`body_mass_g`)**를 봅니다.
  이 두 변수는 **문제 5까지 계속 사용**하며, 상관 → 회귀 → 검정 → 진단으로 이어집니다.

- **피어슨 상관계수(r)**는 관계의 방향(부호)과 강도(절대값)를 −1~+1로 나타냅니다.
  다만 피어슨은 **두 변수가 정규분포라는 전제**에 기대는 모수적 방법입니다.

**🎯 이 문제로 배우는 것**

- 정규성 점검 → 피어슨(r과 p-값) → 스피어만 비교의 순서로, **강의에서 배운 절차 그대로** 상관을 구합니다.

```python
# 문제 2 · 데이터 준비 — 실행만 하세요
df_corr = penguins[['flipper_length_mm', 'body_mass_g']].dropna()

print("결측치 제거 전:", len(penguins), "행")
print("결측치 제거 후:", len(df_corr), "행")

# 두 변수의 관계를 그림으로 먼저 확인 (문제 1의 교훈!)
sns.scatterplot(data=df_corr, x='flipper_length_mm', y='body_mass_g', alpha=0.6)
plt.title("날개 길이와 몸무게의 관계")
plt.xlabel("날개 길이 (mm)")
plt.ylabel("몸무게 (g)")
plt.show()
```

### Q1 · 피어슨의 전제인 정규성을 점검해 봅시다

- `flipper_length_mm`과 `body_mass_g` **각각**에 대해 정규성 검정(Shapiro-Wilk)을 수행하세요.

- 유의수준 0.05를 기준으로 '정규성을 기각하는지'까지 함께 출력하세요.

- `💡 힌트` `stats.shapiro(데이터)`는 통계량과 p-값을 함께 돌려줍니다.

```python
# 문제 2 · Q1
# 여기에 코드를 작성해주세요.

from scipy import stats

alpha = 0.05

for col in ['flipper_length_mm', 'body_mass_g']:
    stat, p = stats.shapiro(df_corr[col])
    result = "정규성 기각 (정규분포 아님)" if p < alpha else "정규성 기각 못함 (정규분포로 볼 수 있음)"
    print(f"{col}: W = {stat:.4f}, p-값 = {p:.2e} → {result}")
```

### Q2 · 피어슨 상관계수와 p-값을 함께 구해 봅시다

- 상관계수만이 아니라 **p-값도 함께** 구하세요. 상관계수도 가설검정의 대상입니다.

- 결과는 반드시 `r`, `p_value`라는 이름으로 저장하세요. **문제 3·4에서 이어서 사용합니다.**

- `💡 힌트` `r, p_value = stats.pearsonr(x, y)` — 귀무가설은 "모집단의 상관계수가 0이다"입니다.

```python
# 문제 2 · Q2
# 여기에 코드를 작성해주세요.

x = df_corr['flipper_length_mm']
y = df_corr['body_mass_g']

r, p_value = stats.pearsonr(x, y)

print(f"피어슨 상관계수 r = {r:.4f}")
print(f"p-값 = {p_value:.2e}")
if p_value < alpha:
    print("→ 귀무가설(모상관계수 = 0) 기각: 상관관계가 통계적으로 유의함")
else:
    print("→ 귀무가설 기각 못함: 상관관계가 유의하지 않음")
```

### Q3 · 스피어만 상관계수를 구해 피어슨과 비교해 봅시다

- 스피어만 상관계수(ρ)를 구하세요.

- Q2의 피어슨 값과 **나란히 출력**해 두 값의 차이를 눈으로 확인하세요.

- `💡 힌트` `stats.spearmanr(x, y)`

```python
# 문제 2 · Q3
# 여기에 코드를 작성해주세요.

rho, p_spearman = stats.spearmanr(x, y)

print(f"피어슨  r   = {r:.4f}  (p-값 = {p_value:.2e})")
print(f"스피어만 ρ  = {rho:.4f}  (p-값 = {p_spearman:.2e})")
print(f"차이 (r - ρ) = {r - rho:.4f}")
```

### Q5 · 상관과 회귀를 잇는 두 관계식을 확인해 봅시다

- 다음 두 식이 실제로 성립하는지 **숫자로** 확인하세요.
  - ① 기울기 = `r` × (y의 표준편차 / x의 표준편차)
  - ② 결정계수 `R²` = `r²` _(단순회귀에서만 성립)_

- 좌변과 우변을 **나란히 출력**해 두 값이 일치하는지 눈으로 확인하세요.

- `💡 힌트` 문제 2 Q2의 `r`을 사용합니다. 표준편차는 `.std(ddof=1)`, 모델의 R²는 `model.rsquared`입니다.

```python
# 문제 3 · Q5
# 여기에 코드를 작성해주세요.

x = penguins_cleaned['flipper_length_mm']
y = penguins_cleaned['body_mass_g']

# ① 기울기 = r × (y의 표준편차 / x의 표준편차)
sd_x = x.std(ddof=1)
sd_y = y.std(ddof=1)

slope_model = model.params['flipper_length_mm']   # 좌변: 회귀 결과표의 기울기
slope_calc = r * (sd_y / sd_x)                     # 우변: r로 계산한 기울기

print("① 기울기 = r × (sd_y / sd_x)")
print(f"   x의 표준편차 = {sd_x:.4f}, y의 표준편차 = {sd_y:.4f}")
print(f"   좌변 (모델의 기울기) = {slope_model:.6f}")
print(f"   우변 (r × sd_y/sd_x) = {slope_calc:.6f}")
print(f"   일치 여부: {np.isclose(slope_model, slope_calc)}")

# ② R² = r²
print("\n② R² = r²")
print(f"   좌변 (모델의 R²) = {model.rsquared:.6f}")
print(f"   우변 (r의 제곱)  = {r**2:.6f}")
print(f"   일치 여부: {np.isclose(model.rsquared, r**2)}")
```

### Q4 · 상관계수의 검정과 같은 검정인지 확인해 봅시다

- 강의에서 배운 공식으로 t-통계량을 **직접 계산**하세요.

  `t = r × √(n − 2) / √(1 − r²)`

- 이 값을 `model.tvalues['flipper_length_mm']`와 **나란히 출력**해 비교하세요.

- `💡 힌트` `n`은 표본 크기(`len(penguins_cleaned)`), `r`은 문제 2 Q2에서 구한 값입니다.

```python
# 문제 4 · Q4
# 여기에 코드를 작성해주세요.

import numpy as np

n = len(penguins_cleaned)
t_calc = r * np.sqrt(n - 2) / np.sqrt(1 - r**2)
t_model = model.tvalues['flipper_length_mm']

print(f"표본 크기 n = {n}")
print(f"공식으로 계산한 t  = {t_calc:.6f}")
print(f"회귀 모델의 t      = {t_model:.6f}")
print(f"일치 여부: {np.isclose(t_calc, t_model)}")
```

### Q5 · p-값이 크게 나오는 사례를 만나 봅시다

- `💡 힌트` 먼저 필요한 열의 결측치를 제거한 뒤 조건으로 걸러 냅니다.
  `df[(df['species'] == 'Adelie') & (df['sex'] == 'Male')]`

```python
# 문제 4 · Q5
# 여기에 코드를 작성해주세요.

# 필요한 열의 결측치 제거 후 Adelie 수컷만 선택
adelie_male = penguins.dropna(subset=['bill_length_mm', 'body_mass_g', 'sex'])
adelie_male = adelie_male[(adelie_male['species'] == 'Adelie') & (adelie_male['sex'] == 'Male')]

model_am = smf.ols(formula='body_mass_g ~ bill_length_mm', data=adelie_male).fit()

p_am = model_am.pvalues['bill_length_mm']
print(f"표본 크기 n = {len(adelie_male)}")
print(f"기울기 = {model_am.params['bill_length_mm']:.2f}")
print(f"기울기의 p-값 = {p_am:.4f}")
print(f"R² = {model_am.rsquared:.4f}")

if p_am < alpha:
    print(f"p-값({p_am:.4f}) < {alpha} → 귀무가설을 기각한다.")
else:
    print(f"p-값({p_am:.4f}) ≥ {alpha} → 귀무가설을 기각하지 못한다.")
```

결정계수 (R²)
반응변수의 전체 변동 중에서 회귀모형이 설명하는 변동의 비율입니다. 0에서 1 사이의 값을 가지며, 1에 가까울수록 모형의 설명력이 높습니다. 1에서 R²를 뺀 나머지는 모형이 설명하지 못한 변동의 비율입니다. 단순회귀에서는 상관계수의 제곱과 같습니다.

예측값 (fitted value)
회귀식에 설명변수 값을 대입하여 계산한 반응변수의 값입니다. 회귀선 위에 놓이는 값입니다.

잔차 (residual)
실제 관측값에서 예측값을 뺀 값으로, 모형이 설명하지 못하고 남은 부분입니다. 양수이면 실제값이 예측보다 크고, 음수이면 실제값이 예측보다 작다는 뜻입니다. 절편을 포함한 최소제곱 회귀에서 잔차의 합과 평균은 항상 0입니다.

잔차 산점도
가로축에 예측값, 세로축에 잔차를 놓고 그린 그래프입니다. 잔차가 0을 기준으로 위아래에 고르게, 일정한 폭으로, 아무런 패턴 없이 흩어져 있으면 모형의 가정이 잘 충족된 것으로 봅니다. 곡선 형태의 패턴은 선형성 가정이 맞지 않음을, 퍼진 폭이 변하는 패턴은 등분산성 가정이 깨졌음을 의미합니다.

등분산성 (homoscedasticity)
설명변수나 예측값의 크기와 관계없이 잔차의 분산(퍼진 정도)이 일정한 상태입니다. 회귀분석의 기본 가정 중 하나입니다.

이분산성 (heteroscedasticity)
예측값의 구간에 따라 잔차의 분산이 달라지는 상태입니다. 잔차 산점도에서 한쪽으로 갈수록 폭이 넓어지거나 좁아지는 깔때기(부채꼴) 모양으로 나타납니다.

이분산성이 미치는 영향
이분산성이 있어도 최소제곱법으로 구한 회귀계수 값 자체는 편향되지 않습니다. 그러나 회귀계수의 표준오차가 부정확하게 계산되어, 이를 바탕으로 하는 t값, p-값, 신뢰구간을 신뢰할 수 없게 됩니다. 즉 문제가 되는 것은 계수 값이 아니라 그 값의 통계적 추론입니다.

브루쉬-페이건 검정 (Breusch-Pagan test)
등분산성을 통계적으로 확인하는 검정입니다. 잔차의 제곱이 설명변수로 설명되는지를 검사합니다.
귀무가설(H₀)은 "등분산이다", 대립가설(H₁)은 "이분산이다"입니다.
p-값이 유의수준보다 작으면 귀무가설을 기각하고 이분산성을 의심하며, p-값이 유의수준 이상이면 귀무가설을 기각하지 못하므로 등분산 가정에 문제가 없다고 판단합니다. 따라서 이 검정에서는 p-값이 클수록 모형의 가정이 잘 지켜진 것입니다.
het_breuschpagan 함수는 LM 통계량, LM p-값, F 통계량, F p-값의 네 값을 반환합니다.

설명변수 행렬 (exog)
모형에 사용된 설명변수 데이터를 행렬로 나타낸 것으로, 절편에 해당하는 상수항 열을 포함합니다. 브루쉬-페이건 검정에서 잔차와 함께 입력값으로 사용됩니다.

회귀계수의 유의성과 모형의 적절성
회귀계수가 통계적으로 유의하다는 것은 설명변수와 반응변수 사이에 선형 관계가 있다는 증거일 뿐, 모형 전체가 적절하다는 뜻은 아닙니다. 좋은 모형인지 판단하려면 계수의 유의성, 설명력(R²), 잔차 진단 결과를 함께 확인해야 합니다.

누락변수
반응변수에 영향을 주지만 모형에 포함되지 않은 변수입니다. 누락변수가 있으면 설명되지 않은 변동이 커지고, 잔차에 패턴이나 이분산성이 나타날 수 있습니다.

다중회귀분석
두 개 이상의 설명변수를 함께 사용하여 반응변수를 설명하는 회귀분석입니다. 범주형 변수는 더미변수로 변환하여 포함합니다. 각 회귀계수는 다른 설명변수를 고정했을 때 해당 변수의 효과를 의미하며, 변수 수가 다른 모형끼리 비교할 때는 수정 결정계수(Adjusted R²)를 사용합니다.

### Q4 · 등분산성을 숫자로 검정해 봅시다

- **브루쉬-페이건 검정**으로 등분산성을 확인하세요.

- H₀는 "등분산이다"입니다. p-값이 0.05보다 작으면 이분산을 의심합니다.

- 판정 결과를 문장으로 함께 출력하세요.

- `💡 힌트` `het_breuschpagan(잔차, model.model.exog)`는
  `(LM통계량, LM p-값, F통계량, F p-값)`을 돌려줍니다.

```python
# 문제 5 · Q4
# 여기에 코드를 작성해주세요.

from statsmodels.stats.diagnostic import het_breuschpagan

lm_stat, lm_p, f_stat, f_p = het_breuschpagan(resid, model.model.exog)

print(f"LM 통계량 = {lm_stat:.4f}, LM p-값 = {lm_p:.4f}")
print(f"F 통계량  = {f_stat:.4f}, F p-값  = {f_p:.4f}")

if lm_p < 0.05:
    print(f"p-값({lm_p:.4f}) < 0.05 → 귀무가설(등분산)을 기각한다. 이분산이 의심된다.")
else:
    print(f"p-값({lm_p:.4f}) ≥ 0.05 → 귀무가설(등분산)을 기각하지 못한다. 등분산 가정에 큰 문제가 없다.")
```
