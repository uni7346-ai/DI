| 옵션 | 의미 | 언제쓰나|
| --- | --- | --- |
| axis=0 | 위아래(행방향)로 쌓기 | 같은 컬럼의 기록을 누적할 때 |
| axis=1 | 좌우(열방향)로 붙이기 | 같은 행에 다른 컬럼을 추가할 때|
| ignore_index=True | 합친 뒤 인덱스를 0부터 다시 매김 | 원래 인덱스가 없미 없을 때(대부분) |
| keys=[...] | 어느 조각에서 왔는지 라벨달기 | 출처를 구분해야 할 때|


| how | 뜻 | 남는 행 | 언제쓰나 |
| --- | --- | --- | --- |
| inner | 교집합 | 양쪽 모두에 있는 키만 | 짝이 맞는 것만보고 싶을 때|
| left | 왼쪽 기분 | 왼쪽은 전부 + 오른쪽은 맞는것만 | 왼쪽 표를 기준으로 정보를 덧붙일 때 |
| right | 오른쪽 기준 | 오른쪽은 전부 | 왼쪽은 맞는 것만 | left의 거울, 자주 안씀 |
| outer | 합집합 | 어느 한쪽에라도 있으면 전부 | 누락을 찾아내고 싶을 때|



| 확인 | 방법 | 무엇을 막나 |
| --- | --- | --- |
| 키 중복 | df["key"].duplicated().sum() | 헹 폭증|
| 짝 안 맞는 키 | ~left["key"],isin(right["key"]) | 조용한 결측|
| 병합 검증 옵션 | merge(..., validated="m:1") | 관계 가정 위반 시 에러로 알려줌 |
| 출처 표시 | merge(..., indicator =True) | 어느 쪽에서 왔는지 _merge 열로 확인|

df.groupby("기준열")["대상열"].집계함수() 형태가 기본

| 패턴 | 코드 예 | 결과|
| --- | --- | --- |
| 한 열, 한 함수 | df.groupby("region")["amount"].mean() | 지열별 평균 금액 |
| 여러 함수 | df.groupby("region")["amount"].agg(["mean", "sum", "count"]) | 지약뱔 퍙균 합 건수|
| 여러 열,이름 지정 | agg(매출=("amount", "sum"), 건수=("order_id", "count")) | 보기 좋은 일므의 요약표 |
| 여러 기준 | df.groupby(["region", "membership"]) | 지역 x 등급 교차 그룹 |

|  | agg | transform |
| --- | --- | --- |
| 결과 행 수 | 그룹 수만큼(줄어듦) | 원본과 동일(안줄어둚) |
| 주 용도 | 요약표 만들기 | 원본에 그룹 통계 컬럼 추가 |
| 대표 활용 | 보고서,대시보드 | 그룹 내 비중,편차, 정규화 |

long : 한 행이 하나의 관측/조합, 데이터 저장,집계에 유리(groupby 결과의 기본모양)

wide : 한 축을 열로 펼치고 사람이 보고 비교하기 좋음(보고서,엑셀 표)

| 도구 | 방향 | 핵심 인자 |
| --- | --- | --- |
| pivot_table | long > wide | index(행), columns(열), values(값), aggfunc(집계함수) |
| melt | wide > long | id_var(고정 열), value_vars(녹일 열) |
| unstack | 인덱스 > 열 | 멀티인덱스의 한 레벨을 열로 |
| stack | 열 > 인덱스 | unstack 의 반대 |

pivot_table은 사실 groupby + unstack을 한번에 해주는 도구
