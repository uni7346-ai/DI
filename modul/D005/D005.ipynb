공백 제거(strip) > 대소문자 통일 (lower/upper) > 표기 치환(replace)

Pandas에서 문자열 컬럼(series)을 다룰 때는 .str을 붙인다. 그러면 파이썬 문자열 기능을 컬럼 전체에 한 번에 적용할 수 있다.

| 기능 | 하는 일 | 예시 |
| --- | --- | --- |
| .str.strip() | 앞뒤 공백 제거 | " 서울 " > "서울" |
| .str.lower()/.str.upper() | 소문자/대문자 통일 | "APP" > "app" |
| .str.replace(a, b) | a를 b로 치환 | "Seoul" > "서울" |
| .str.len() | 글자 수 | "서울" > 2 |

| 기능 | 하는 일 | 결과 예시 |
| --- | --- | --- |
| .str.contains("gmail") | 그 글자를 포함하는가? | True/False |
| .str.startwith("010") | 그 글자로 시작하는가? | True/False |
| .str.split("0") | 기준 글자로 쪼개 리스토로 | ["user001", "gmail.com"] |
| .str.split("@").str[1] | 쪼갠 뒤 N번째 조각 선택 | "gmail.com |
| .str[0:3] | 글자 위치로 잘라내기 | "010" |


| 기호 | 뜻 | 예 |
| --- | --- | --- |
| \d | 숫자 한 개(0~9) | \d\d > "12" |
| \w | 글자/숫자/밑줄 한개 | \w+ > "user001" |
| + | 바로 앞 패턴을 1번 이상 반복 | \d+ > "12345" |
| [...] | 괄호 안 글자 중 하나 | [A-Z] > 대문자 한 개 |
| . | 아무 글자 한개 | |
| () | 뽑아낼 부분을 묶음(그룹) | (\d+) |

-----------------------------------

| 함수 | 하는 일 | 
| --- | ---|
| .str.extract(r"(패턴)") | 그룹 () 부분을 뽑아 새 컬럼으로 |
| .str.findall(r"패턴") | 일치하는 모든 조각을 리스트로 |
| .str.replace(r"패턴". "", regex=True) | 패턴에 맞는 부분을 치환/삭제 |
| .str.contains(r"패턴", regex=True) | 패턴이 잇는지 참/거짓|


| 옵션 | 하는 일 | 언제 |
| --- | --- | --- |
| pd.to.datetime(s) | 알아서 형식을 추론해 변환 | 형식이 깔끔할 때 |
| foramt="..." | 형식을 직접 지정 (가장 정확,빠름) | 형식을 알 때 |
| format="mixed" | 행마다 형식이 다를 때 각자추론 | 포맷이 섞였을 때 |
| errors="coerce" | 변환 못하는 값은 NaT(결측 날짜)로 | 오류값이 섞였을 때|

-------------------------------------------

| 기호 | 뜻 | 예 |
| --- | ---|  --- |
| %Y | 4자리 연도 | 2025 |
| %m | 2자리 월 | 01 |
| %d | 2자리 일 | 12 |
| %H:%M:%S | 시:분:초 | 10:15:32|

| 코드 | 하는 일 |
| --- | --- |
| df.set_index("order_dt") | 날짜를 인덱스 (시간축)로 |
| .resample("D").sum() | 하루 단위로 묶어 합계 ("D"=일) |
| .resample(:W:).sum() | 한 주 단위로 묶어 합계 ("W"=주) |
| .resample("h").count() | 한 시간 단위로 묶어 개수 ("h"=시간) |
| .rolling(7).mean() | 최근 7개의 이동평균 |

rolling() : 지정한 크기의 이동 창을 데이터 위에서 한 칸씩 이동시키며 연산 할 수 있게 하느 메서드
