데이터를 구하는 방법

| 방법 | 설명 | 언제쓰나 | 난이도 |
| --- | --- | ---- | ---- |
| 파일 내려받기 | CSV,Excel,공공데이터 포털 | 이미 정리된 데이터가 공개 돼 있을 때 | 쉬움 |
| API 호출 | 사이트가 공식 제공하는 데이터 창구 | 공식 API가 있을 때 | 중간 |
| 웹 스크래핑 | 웹 페이지의 HTML에서 직접 추출 | API가 없고, 화면에만 데이터가 있을 때 | 중간-높음 |

요청의 구성

| 요소 | 뜻 | 예 |
| --- | --- | ---|
| Method | 무엇을 하려는가 | GET(가져오기), POST(보내기) |
| URL | 어디에 | https://books.toscrape.com/ |
| headers | 부가정보 | User-Agent |

응답의 상태코드
| 코드 | 뜻 | 의미 |
| --- | --- | --- |
| 2xx | 성공 | 200 OK -잘받았다 | 
| 3xx | 이동 | 301 - 주소가 바뀌었다 | 
| 4xx | 내 잘못 | 404 없는 페이지, 403 접근 거부 |
| 5xx | 서버 잘못 | 500 서버오류 |


정적(static) : 서버가 데이터깢 ㅣ채워진 완성 HTML을 보냅다. requsets로 받으면 데이터가 그대로 들어 있어 바로 BeautifulSoup으로 추출 가능하다

동적(dynamic) : 서버가 빈 뼈대만 보내고, 브라우저에서 javaScript가 실행되며 데이터르 채운다. requests는 JS를 실행하지 않으므로 빈 뼈대만 받는다 > BeautifulSoup만으론 안된다.

| 신호 | 정적일 가능성 | 동적일 가능성 |
| --- | ---| ---|
| 브라우저 "페이지 소스보기"에 데이터가 있다 " | O |  |
| 소스엔 없는데 화면엔 보임 |  | O |
| requests로 받은 HTML에 값이 들어있음 | O |  |
| 무한 스크롤"더보기" 클릭으로 내용이 늘어남 |  | O |

선택자와 추출 메서드

| 개념 | 표기 | 예 |
| --- | --- | --- |
| 태그 | 이름 그대로 | div, span, a |
| 클래스 선택 | 앞에. | .price > class가 price인 것 |
| 아이디 선택 | 앞에 # | #product-list > id가 product-list인 것 |
| 계층 선택 | 공백으로 이어 | .card .name > card 안의 name |

------------------------------------------------

| BeautifulSoup 메서드 | 하는일 |
| --- | --- |
| soup.select(선택자) | 선택자에 맞는것을전부 리스트로 (CSS 선택자) |
| soup.select_one(선택자) | 맞는 것 중 첫 번째 하나 |
| soup,find_all("태그") | 해당 태그를 전부 |
| .get_text() | 태그 안의 글자만 꺼내기 |
| tag("속성") 또는 tag.get("속성") | 속성값 꺼내기 |

도구 지형과 선택 기준

| 도구 | 한마디 | JS실행 | 속도 | 언제 |
| --- | --- | --- | --- | --- |
| requests +BeautifulSoup | 정적 HTML 파싱 | x | 빠름 | 정적 페이지(오늘) |
| Selenium | 브라우저 자동화(원조) | o | 느림 | 동적, 구형프로젝트 |

수집 전 4가지 점검

| 점검 | 무엇을| 어떻게 |
| --- | --- | ---|
| robust.txt | 사이트가 허용/금지한 경로 | 사이트/robots.txt 확인, 코드로 검사 |
| 이용약관(ToS)저작권 | 데이터 재사용, 재배포 가능 여부 | 약관 읽기. 수집 가능 ≠ 재배포 가능 |
| 요청 예의 (rate limit) | 서버에 부담 안 주기 | 요청 사이 time.sleep(), 동시 요청 자제 |
| 개인정보 | 개인 식별 데이터 | 수집, 저장 최소화, 가능하면 회피 |
