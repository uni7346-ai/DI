series.map :dict 또는 1 - 입력함수  한값 > 다른 값으로 치환
series.apply : 1 - 입력함수 값마다 사용자 정의 함수 적용
DataFrame.apply : 1 - 입력함수  행 또는 열 단위변환
벡터화 : 연산자 내장 메서드  컬럼 전체에  한번에

Label Encoding : 각 카테고리에 정수 한개씩(트리계열 모델, 순서가 진짜 있을 때)
ordinal Encoding : 사용자가 정한 순서대로 정수 매핑(순서가 명확한 카테고리)
One-Hot Encoding : 카테고리마다 0/1 컬럼 한 개씩(순서없는 카테고리)

Min - Max : (x - min) / (max - min) 범위 : [0,1]
standard : (x- μ) / σ 범위 : [-3,3]
Robust : (x-median) / IQR 범위 : 분포의존

dropna(subset=[...]) : 특정 열의 결측행 제거
fillna({col: val}) : 결측 채우기
query("조건식") : 문자열 조건으로 필터
assign(new=...) : 새 컬럼 추가
rename(columns={...}) : 컬럼명 바꾸기
astype({col: dtype}) : 자료형 변환
sort_value(by=...) : 정렬
reset_index(drop=True) : 인덱스 리셋
drop_duplicates() : 중복 행 제거g

| 도구 | 무엇을 받나 | 어디에 쓰나 | 대표형태 |
| --- | --- | --- | --- |
| Series.map | dict또는 1-입력함수 | 한 값 > 다른 값으로 치환 | s. map({"M":1,"F":0}) |
| Series.apply | 1-입력함수 | 값마다 사용자 정의 함수 적용 | s.apply(lambda x: x*1.1) |
|DataFrame.apply | 1-입력함수 | 행 또는 열 단위 변환 | df.apply(func,axis=1) |
| 벡터화 | (함수없이)연산자 내장메서드 | 컬럼 전체에 한 번에 | s * 1.1, s.str.lower()|

| 인코딩 | 어떻게 바꾸나 | 어울리는 상황 | 주의점|
|---|---|---|---|
|Label Encoding | 각 카테고리에 정수 한 개씩(임의 순사)| 트리계열 모델, 순서가 진짜 있을 때 | 선형/거리 모델에 쓰면 가짜 순서학습|
|Ordinal Encoding | 사용자가 정한 순서대로 정수 매핑 | 순서가 명확한 카테고리(등급, 학년) | 순서를 직접 지정해야 함 | 
One-Hot Encoding | 카테고리마다 0/1 컬럼 한 개씩| 순서 없는 카테고리 | 카디널리티 (고유값 수)가 크면 컬럼폭발 |
